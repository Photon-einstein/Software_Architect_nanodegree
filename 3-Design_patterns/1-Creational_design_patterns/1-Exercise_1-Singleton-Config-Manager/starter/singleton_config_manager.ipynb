### Exercise: Singleton Configuration Manager  

**Goal:** Implement a Singleton `ConfigManager` class to manage application-wide settings.  

**Steps:**  
1. Create a class `ConfigManager`.  
2. Ensure that only **one instance** of `ConfigManager` can exist.  
3. Add a dictionary `settings` to store key-value pairs.  
4. Implement methods:  
   - `set(key, value)` → stores a setting.  
   - `get(key)` → retrieves a setting.  

**Behavior to Verify:**  
```python
config1 = ConfigManager()
config2 = ConfigManager()

config1.set("theme", "dark")
print(config2.get("theme"))   # "dark" -  Both share same settings

print(config1 is config2)     # True - Only one instance

````python
# Exercise: Implement a Singleton ConfigManager class.
# The ConfigManager should:
#   - Only allow ONE instance to be created (Singleton).
#   - Store key-value settings in a dictionary.
#   - Provide `set(key, value)` and `get(key)` methods.
#   - Raise KeyError if `get(key)` is called for a missing key.


class ConfigManager:
    _instance = None  # Class variable to store the one instance

    def __new__(cls, *args, **kwargs):
        # Implement Singleton behavior
        # If _instance is None, create it and assign to cls._instance
        # Otherwise, return the existing cls._instance
        if cls._instance is None:
            print("No instance exists yet")
            cls._instance = super(ConfigManager, cls).__new__(cls)
        return cls._instance

    def __init__(self):
        # Initialize settings dictionary only once
        # (Hint: Use hasattr(self, "settings") to avoid re-initialization)
        if not hasattr(self, "settings"):
            self.settings = {}

    def set(self, key, value):
        # Save key-value pair to settings dictionary
        self.settings[key] = value

    def get(self, key):
        # Retrieve the value for a key
        # If key is not found, raise KeyError with a helpful message
        if key not in self.settings:
            raise KeyError(f"Settings '{key}' not found.")
        return self.settings[key]
````

````python
import unittest


class TestSingletonConfigManager(unittest.TestCase):

    def test_single_instance(self):
        config1 = ConfigManager()
        config2 = ConfigManager()
        self.assertIs(config1, config2, "ConfigManager is not a Singleton!")

    def test_set_and_get(self):
        config = ConfigManager()
        config.set("theme", "dark")
        self.assertEqual(
            config.get("theme"), "dark", "Stored value not retrieved correctly"
        )

    def test_shared_state(self):
        config1 = ConfigManager()
        config2 = ConfigManager()
        config1.set("language", "English")
        self.assertEqual(
            config2.get("language"), "English", "Instances do not share state"
        )


if __name__ == "__main__":
    unittest.main(argv=["first-arg-is-ignored"], exit=False)
````
