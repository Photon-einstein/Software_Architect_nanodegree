### Exercise: House Builder Pattern

**Goal:** Implement the Builder Pattern to construct `House` objects step by step.  
A `House` can have optional components: `walls`, `roof`, `windows`, `doors`, `garage`.

**Steps:**  
1. Create a `House` class that stores its components and has a method `show_details()` that returns a string listing all components.  
2. Create a `HouseBuilder` class with methods to add each component:  
   - `set_walls(type)`  
   - `set_roof(type)`  
   - `set_windows(count)`  
   - `set_doors(count)`  
   - `set_garage(type)`  
   Each method should return `self` for **method chaining**.  
3. Create a `build()` method that returns the final `House` object.  

**Behavior to Verify:**  
```python
builder = HouseBuilder()
house = builder.set_walls("Brick").set_roof("Tile").set_windows(4).build()
print(house.show_details())
# Output: Walls: Brick, Roof: Tile, Windows: 4, Doors: None, Garage: None


````python
# Implement the House class
class House:
    def __init__(self):
        # Store walls, roof, windows, doors, garage
        self.walls = None
        self.roof = None
        self.windows = None
        self.doors = None
        self.garage = None

    def show_details(self):
        return (
            f"Walls: {self.walls}, Roof: {self.roof}, Windows: {self.windows}, "
            f"Doors: {self.doors}, Garage: {self.garage}"
        )


# TODO: Implement the HouseBuilder class
class HouseBuilder:
    def __init__(self):
        # Initialize a House object
        self._house = House()

    def set_walls(self, walls_type):
        # Add walls to house
        # Return self for chaining
        self._house.walls = walls_type
        return self

    def set_roof(self, roof_type):
        # Add roof to house
        # Return self
        self._house.roof = roof_type
        return self

    def set_windows(self, count):
        # Add number of windows
        # Return self
        self._house.windows = count
        return self

    def set_doors(self, count):
        # Add number of doors
        # Return self
        self._house.doors = count
        return self

    def set_garage(self, garage_type):
        # Add garage type
        # Return self
        self._house.garage = garage_type
        return self

    def build(self):
        # Return the final House object
        return self._house
````

````python
import unittest


class TestHouseBuilder(unittest.TestCase):

    def test_build_house(self):
        builder = HouseBuilder()
        house = builder.set_walls("Brick").set_roof("Tile").set_windows(4).build()
        self.assertEqual(
            house.show_details(),
            "Walls: Brick, Roof: Tile, Windows: 4, Doors: None, Garage: None",
        )

    def test_chaining_all_components(self):
        builder = HouseBuilder()
        house = (
            builder.set_walls("Concrete")
            .set_roof("Shingle")
            .set_windows(6)
            .set_doors(2)
            .set_garage("Double")
            .build()
        )
        self.assertEqual(
            house.show_details(),
            "Walls: Concrete, Roof: Shingle, Windows: 6, Doors: 2, Garage: Double",
        )


if __name__ == "__main__":
    unittest.main(argv=["first-arg-is-ignored"], exit=False)
````
